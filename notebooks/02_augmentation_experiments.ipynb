# Experiment B — QA augmentation robustness

Проверка двух видов текстовых преобразований; описательные метрики без ложного заявления о сохранении смысла

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

a=pd.read_csv(Path.cwd()/'reports/augmentations.csv')
display(a.groupby('method').agg(n=('original_id','size'), cosine_mean=('cosine','mean'),length_delta_mean=('length_delta','mean')).round(3))
```

```python
a.boxplot(column='cosine',by='method',figsize=(8,4));plt.suptitle('');plt.title('Embedding shift under augmentation');plt.show()
```
