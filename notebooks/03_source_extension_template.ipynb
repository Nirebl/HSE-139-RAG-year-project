```python
import json
from pathlib import Path

import pandas as pd

rows=json.loads((Path.cwd()/'data/processed/documents.json').read_text(encoding='utf-8'))
df=pd.DataFrame(rows)
display(df.groupby('source_type').agg(n=('document_id','size'),unique_ids=('document_id','nunique')))
print('Требуемые поля:',list(df.columns))
```

```python
def validate_new_rows(rows):
    required={'document_id','source_type','source_name','source_url','content'}
    errors=[]
    for i,row in enumerate(rows):
        if not required.issubset(row):errors.append((i,'missing fields'))
        if not isinstance(row.get('content'),str) or not row.get('content','').strip():errors.append((i,'empty content'))
    return errors
print('Базовая проверка первых 10 записей:',validate_new_rows(rows[:10]))
```
