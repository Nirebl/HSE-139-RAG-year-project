# Experiment A — BM25 vs dense: candidate agreement



```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

p=Path.cwd()/'reports/retrieval_candidates.csv'
df=pd.read_csv(p)
print('Queries:',df.query_id.nunique(), 'pairs:',len(df))
display(df.groupby('source_type').overlap_at3.agg(['count','mean','median']).round(3))
```

```python
df.boxplot(column='overlap_at3',by='source_type',figsize=(8,4));plt.suptitle('');plt.title('Top-3 overlap: BM25 vs Dense (not quality)');plt.ylabel('Overlap@3');plt.show()
```
